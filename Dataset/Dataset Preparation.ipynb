```python
import numpy as np
import pandas as pd
```

### Preparing The Training Dataset

#### Code Before Review

```python
with open('Dataset/train/src-train.txt', 'r', encoding = 'utf-8-sig') as file:
    lines = file.readlines()
```

```python
oc = list()
for line in lines:
    oc.append(line.strip())
    
oriCode = pd.DataFrame(data=oc, columns=['OriginalCode'])
```

```python
oriCode
```

```python
len(oriCode)
```

#### Code Implemented Based on The Review Comment

```python
with open('Dataset/train/tgt-train.txt', 'r', encoding = 'utf-8-sig') as file:
    lines = file.readlines()
```

```python
rc = list()
for line in lines:
    rc.append(line.strip())
    
revCode = pd.DataFrame(data=rc, columns=['ReviewedCode'])
```

```python
revCode
```

### Merging OriginalCode and ReviewedCode into a single pandas DataFrame

```python
data_train = pd.DataFrame(pd.concat([oriCode, revCode], axis=1))
```

```python
data_train.head()
```

### Preparing the validation data set

```python
with open('Dataset/eval/src-val.txt', 'r', encoding = 'utf-8-sig') as file:
    lines = file.readlines()
    
oc_v = list()
for line in lines:
    oc_v.append(line.strip())
    
oriCode_v = pd.DataFrame(data=oc_v, columns=['OriginalCode'])
```

```python
oriCode_v
```

```python
with open('Dataset/eval/tgt-val.txt', 'r', encoding = 'utf-8-sig') as file:
    lines = file.readlines()
    
rc_v = list()
for line in lines:
    rc_v.append(line.strip())
    
revCode_v = pd.DataFrame(data=rc_v, columns=['ReviewedCode'])
```

```python
revCode_v
```

```python
data_val = pd.DataFrame(pd.concat([oriCode_v, revCode_v], axis=1))
```

```python
data_val
```

### Merging The Training and Validating sets into one dataframe

```python
final_train = pd.DataFrame(pd.concat([data_train, data_val]))
```

```python
final_train
```

```python
# final_train.to_csv("final_train", sep=',', index=False, encoding='utf-8')
```

```python
import json
```

```python
ftrain = final_train.to_dict(orient="records")

with open("train.jsonl", "w") as f:
    for line in ftrain:
        f.write(json.dumps(line) + "\n")
```
